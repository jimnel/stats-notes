```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
theta = 0.3

def sample(n):
    return np.random.choice([0, 1], n, p=[1-theta, theta])

def prob(m, param):
    return m*param + (1-m)*(1-param)
```

```python
samples = sample(1000)

params = np.linspace(0., 1., 101)

priors = np.ones_like(params)
priors /= np.sum(priors)

for s in samples:
    #print(s)
    posterior = prob(s, params)
    #print(posterior)
    priors = posterior * priors / sum(priors * posterior)

plt.plot(params, priors)
plt.axvline(theta, ls='--', c='k')
```
